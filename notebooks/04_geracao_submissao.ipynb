# 04 — Geração da submissão / Submission generation

**PT:** Esta etapa usa o modelo salvo para prever a base de teste e cria `resultados.csv` na raiz do projeto, no formato esperado pela competição.

**EN:** This step uses the saved model to predict the test set and creates `resultados.csv` at the project root, in the competition's expected format.

**Pré-requisitos / Prerequisites:** execute notebooks 02 and 03 first.

```python
from pathlib import Path

import joblib
import pandas as pd

PROJECT_DIR = Path('..')
test = pd.read_csv(PROJECT_DIR / 'DATA/processed/test_prepared.csv')
artifact = joblib.load(PROJECT_DIR / 'artifacts/titanic_model.joblib')

model = artifact['model']
features = artifact['features']
```

## Checagem das entradas / Input check

**PT:** Antes da previsão, verificamos se as colunas do teste correspondem exatamente às usadas no treino.

**EN:** Before prediction, we check that test columns exactly match those used during training.

```python
X_test = test[features]
assert list(X_test.columns) == features
predictions = model.predict(X_test)

submission = pd.DataFrame({
    'PassengerId': test['PassengerId'],
    'Survived': predictions,
})
submission.head()
```

## Exportação / Export

**PT:** `index=False` impede que o índice interno do pandas seja incluído no CSV. Revise as primeiras linhas e envie o arquivo ao Kaggle quando estiver satisfeito.

**EN:** `index=False` prevents pandas' internal index from being included in the CSV. Review the first rows and upload the file to Kaggle when satisfied.

```python
output_path = PROJECT_DIR / 'resultados.csv'
submission.to_csv(output_path, index=False)
print(f'Submissão salva / Submission saved: {output_path.resolve()}')
submission['Survived'].value_counts().rename_axis('Survived').to_frame('count')
```
