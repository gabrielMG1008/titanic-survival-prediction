```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

DATA_DIR = Path('../DATA')
train = pd.read_csv(DATA_DIR / 'train.csv')
test = pd.read_csv(DATA_DIR / 'test.csv')

print(f'Treino / Training: {train.shape}')
print(f'Teste / Test: {test.shape}')
```

## Visão inicial / First look

**PT:** `Survived` existe apenas na base de treino e é a variável-alvo: `1` significa que o passageiro sobreviveu.

**EN:** `Survived` exists only in the training set and is the target variable: `1` means the passenger survived.

```python
train.info()
test.info()
```

## Valores ausentes / Missing values

**PT:** A próxima tabela define o que será tratado na preparação. A imputação será calculada a partir do treino, evitando que informações do teste influenciem o modelo.

**EN:** The next table determines what will be handled during preparation. Imputation will be learned from the training data, preventing test-set information from influencing the model.

```python
missing = pd.DataFrame({
    'train_missing': train.isna().sum(),
    'test_missing': test.isna().sum(),
})
missing[missing.any(axis=1)].sort_values('train_missing', ascending=False)
```

## Análise inicial da sobrevivência / Initial survival analysis

**PT:** Os gráficos são exploratórios: ajudam a formular hipóteses, mas não substituem a validação do modelo.

**EN:** These plots are exploratory: they help formulate hypotheses but do not replace model validation.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
train.groupby('Sex')['Survived'].mean().plot.bar(ax=axes[0])
train.groupby('Pclass')['Survived'].mean().plot.bar(ax=axes[1])
train.assign(FamilySize=train.SibSp + train.Parch).groupby('FamilySize')['Survived'].mean().plot.bar(ax=axes[2])
axes[0].set_title('Sexo / Sex')
axes[1].set_title('Classe / Class')
axes[2].set_title('Tamanho da família / Family size')
plt.tight_layout()
```
